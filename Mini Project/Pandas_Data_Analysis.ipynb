# **Create a DataFrame**

```python
import pandas as pd
student = {
    "Name" : ["Alice", "John", "Julia"],
    "Age" : [20, 21, 22],
    "Mark" : [85, 90, 75]
}

df = pd.DataFrame(student)
print(df)
```
