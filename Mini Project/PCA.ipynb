**Step-2 : Import Libraries**

```python
#import required libraries
import numpy as np
import pandas as pd
from sklearn.decomposition import PCA
import matplotlib.pyplot as plt
import gradio as gr
```

**Step-3 : PCA Compression Function**

```python
def compress_image(image, n_component):
  #convert the image into Numpy array
  image = np.array(image)

  #-----------------------------------
  #RED CHANNEL
  #-----------------------------------
  #Get the Red channel from the image
  red = image[:, :, 0]

  #Create a PCA model for the Red channel
  pca_red = PCA(n_components = n_component)

  #Compress the Red channel
  red_compressed = pca_red.fit_transform(red)

  #Reconstruct the Read Channel
  red_reconstructed = pca_red.inverse_transform(red_compressed)

  #-----------------------------------
  #GREEN CHANNEL
  #-----------------------------------
  #Get the green channel from the image
  green = image[:, :, 1]

  #Create a PCA model for the green channel
  pca_green = PCA(n_components = n_component)

  #Compress the Green channel
  green_compressed = pca_green.fit_transform(green)

  #Reconstruct the green Channel
  green_reconstructed = pca_green.inverse_transform(green_compressed)

  #-----------------------------------
  #BLUE CHANNEL
  #-----------------------------------
  #Get the Blue channel from the image
  blue = image[:, :, 2]

  #Create a PCA model for the Blue channel
  pca_blue = PCA(n_components = n_component)

  #Compress the Blue channel
  blue_compressed = pca_blue.fit_transform(blue)

  #Reconstruct the Blue Channel
  blue_reconstructed = pca_blue.inverse_transform(blue_compressed)

  #------------------------------------
  #COMBINE RGB CHANNELS
  #------------------------------------
  #Combine the three
  compressed_image = np.stack((red_reconstructed, green_reconstructed, blue_reconstructed), axis = 2)

  #Keep pixel value between 0 to 255
  compressed_image = np.clip(compressed_image).astype(np.uint8)

  #Convert the image to unsigned 8-bit format
  # compressed_image = compressed_image.astype(np.uint8)

  return compressed_image

```

**Step-4 : Create Gradio UI**

```python
#Create a Gradio Block application
with gr.Blocks(title = "Image Compress Using PCA") as app:

  #Add the appplication title
  gr.Markdown("# 📊Image Compression Using PCA")

  #Add a short description
  gr.Markdown("Upload an image and compress it using PCA")

  #Create a row for input and output
  with gr.Row():

    #------------------------
    #INPUT SECTION
    #------------------------
    with gr.Column():
      #Add an image upload component
      input_image = gr.Image(
          type = "numpy",
          label = "Orginal Image"
      )

      #Add a Slider for PCA components
      components = gr.Slider(
          minimum = 10,
          maximum = 100,
          value = 50,
          step = 10,
          label = "Number of PCA components"
      )

      #Add a button for compression
      button = gr.Button("Compress Image")

    #-----------------------
    #OUTPUT SECTION
    #-----------------------
    #Create a column for output

    with gr.Column():
      #Add an output image component
      output_image = gr.Image(
          label = "Compressed Color Image"
      )

      #-----------------------------
      #Connect Button to Function
      #-----------------------------
      button.click(
          fn = compress_image,
          inputs = [input_image, components],
          outputs = output_image
      )

  #launch
  app.launch(debug = True)
```
