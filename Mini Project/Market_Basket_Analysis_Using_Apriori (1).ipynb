Step-2 : import Libraries

```python
import numpy as np
import pandas as pd
import gradio as gr

from mlxtend.preprocessing import TransactionEncoder
from mlxtend.frequent_patterns import apriori, association_rules
```

**Step-3 : Create Sample Transaction Data**

```python
#Each list represents one customer's shopping basket
sample_transactions = [
    ["Milk", "Bread", "Butter"],
    ["Bread", "Butter"],
    ["Milk", "Bread"],
    ["Milk", "Bread", "Butter", 'Eggs'],
    ["Bread", "Eggs"],
    ["Mike", "Eggs"],
    ["Mile", "Bread", "Butter"],
    ["Bread", "Butter", "Eggs"],
    ["Mile", "Bread", "Eggs"],
    ["Mike", "Butter"]
]
```

**Step-4 : Convert Transactions to DataFrame**

```python
def create_dataframe(transactions):

  #Create a TaransactionEncoder
  encoder = TransactionEncoder()

  #Convert transactions into True/False values.
  encoded = encoder.fit(transactions).transform(transactions)

  #Craete a pandas dataframe
  dataframe = pd.DataFrame(
      encoded,
      columns = encoder.columns_
  )

  return dataframe

#Create the default dataset
df = create_dataframe(sample_transactions)
```

**Step-5 : Format Association Rules**

```python
def format_rules(rules):
  results = []

  #Fo through every rules
  for _, row in rules.iterrows():

    #Convert product into text
    antecedent = ", ".join(row["antecedents"])
    consequent = ", ".join(row["consequents"])

    #Create one rwo result
    results.append({
        "If Costomer Buys" : antecedent,
        "Then Customer May Buy" : consequent,
        "Support" : round(row["support"], 3),
        "Confidence" : round(row["confidence"], 3),
        "Lift" : round(row["lift"], 3)
    })

  #Crate results to a DataFrame
  return pd.DataFrame(results)



```

**Step-6 : Upload CSV Dataset**

```python
def upload_dataset(file):
  global df

  #if the user does not upload a file, use the default sample dataset
  if file is None:

    return("No File upload, using sample dataset", df)

  try:

    #Read the uploaded CSV file.
    uploaded_data = pd.read_csv(file)

    #Create an empty list for transactions
    transactions = []

    for _, row in uploaded_data.iterrows():
      #Create a list of products in this transactions
      items = []

      for item in row:
        #Ignore empty cells.
        if pd.notna(item):
          item = str(item).strip()
          if item != "":
            items.append(item)

      #Add the transactions if it contains products
      if len(items) > 0:
        transactions.append(items)

    #Check the dataset contains transactions
    if len(transactions) == 0:
      return("The uplaod dataset is empty.", pd.DataFrame())


    #Convert Transactions to DataFrame
    df = create_dataframe(transactions)

    return (f"Dataset uploaded successfully"
            f"{len(transactions)} tarnsactions found", df)

  except Exception as error:
    return(
        f"Error reading dataset : {error}", pd.DataFrame()
    )
```

**Step-7 : Run Apriori Algorithm**

```python
def run_apriori(min_support, min_confidence):
  global df

  #Check if the dataset is empty
  if df is None or df.empty:
    return(
        "Dataset is empty.",
        pd.DataFrame(), #itemset_outputs
        pd.DataFrame() #rules_outputs
    )


  try:
    #-------------------------------------------------
    #Step-1 : Final Frequent Itemset
    #-------------------------------------------------

    frequent_itemsets = apriori(
        df,
        min_support = min_support,
        use_colnames = True
    )

    #Check if frequent itemsets were found
    if frequent_itemsets.empty:
      return(
          "No frequent itemsets found. Try lowering Minium Support",
          pd.DataFrame(),
          pd.DataFrame()
      )

    #---------------------------------------------------
    #Step-2 : Find Association Rules
    #---------------------------------------------------
    rules = association_rules(
        frequent_itemsets,
        metric = "confidence",
        min_threshold = min_confidence
    )

    #---------------------------------------------------
    #Step-3 : Format Frequent itemsets
    #---------------------------------------------------
    itemsets_output = frequent_itemsets.copy()

    itemsets_output["itemsets"] = itemsets_output["itemsets"].apply(
        lambda items: ",".join(list(items)) #Milk, Bread
    )

    #Round support values
    itemsets_output["support"] = itemsets_output["support"].round(3)

    #---------------------------------------------------
    #Step-4 : Format Association Rules
    #---------------------------------------------------
    if rules.empty:
      message = (
          f"Found {len(frequent_itemsets)}"
          f"Frequent itemsets, but no association"
          f"Rules were found. Try Lowering Minium Confidence"

      )

      return(
          message,
          itemsets_output,
          pd.DataFrame()
      )

    #----------------------------------------------------
    # Step-5 : Format Rules
    #----------------------------------------------------
    rules_outputs = format_rules(rules)

    #Create the message
    message = (
        f"Found {len(frequent_itemsets)}"
        f"frequent itemsets and"
        f"{len(rules)} association rules"
    )

    return(
        message,
        itemsets_output,
        rules_outputs
    )

  except Exception as error:
    return(
        f"Error reading dataset : {error}",
        pd.DataFrame(),
        pd.DataFrame()
    )
```

**Step-8 : Create Gradio UI**

```python
with gr.Blocks() as app:
  gr.Markdown("# 🛒Market Basket Analysis Using Apriori")
  gr.Markdown("This application uses the **Apriori Algorithm** to discover relationship between products")
  gr.Markdown(" **Example**")
  gr.Markdown("If a customer buys **Mike** → they may also buy **Bread**")

  #Dataset section
  gr.Markdown("## 📂Dataset")
  gr.Markdown("Upload a CSV file containing transaction data")

  with gr.Row():
    dataset_upload = gr.File(
        label = "Upload Transaction CSV",
        file_types = [".csv"],
        type = "filepath"
    )
    upload_button = gr.Button("🔺Load Dataset")

  #Dataset Status
  upload_message = gr.Textbox(
      label = "Dataset Status"
  )

  #Dataset Preview
  dataset_preview = gr.DataFrame(
      label = "Dataset View",
      interactive = True
  )

  #Apriori Settings
  with gr.Row():
    #Minimu support
    min_support = gr.Slider(
        minimum = 0.01,
        maximum = 1.0,
        value = 0.3,
        label = "Minimum Support"
    )

    #Minimum confidence
    min_confidence = gr.Slider(
        minimum = 0.01,
        maximum = 1.0,
        value = 0.5,
        label = "Minimum Confidence"
    )


  predicted_buttton = gr.Button("🔎Find Association Rules")

  #Output Section
  #Result Message
  message = gr.Textbox(
      label = "Result"
  )

  #Frequent Itemsets
  gr.Markdown("📂Frequent Itemsets")
  itemsets_output = gr.DataFrame(
      label = "Frequent Itemsets",
      interactive = False
  )

  #Association Rules
  gr.Markdown("## ♟️Association Rules")
  rules_output = gr.DataFrame(
      label = "Association Rules",
      interactive = False
  )

  #Upload Button Event
  upload_button.click(
      fn = upload_dataset,
      inputs = dataset_upload,
      outputs = [upload_message, dataset_preview]
  )

  #Apriori Button Event
  predicted_buttton.click(
      fn = run_apriori,
      inputs = [min_support, min_confidence],
      outputs = [message, itemsets_output, rules_output]
  )

#launch
app.launch(debug = True)



```
