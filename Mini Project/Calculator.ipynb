```python
 #install gradio
!pip install gradio -q

#import gradio
import gradio as gr

#define function
def calculate(num1, num2):
  return num1+num2

#Create interface
app = gr.Interface(
    fn = calculate,
    inputs = [gr.Number(label = "Number 1", placeholder="Enter the First Number"), gr.Number(label = "Number 2", placeholder="Enter the Second Number")],
    outputs = gr.Number(label = "Result"),
    title = "Calculate App",
    description = "Enter numbers to calculate"
)

app.launch(share = True)
```

```python
#Test

def operator(num1, choice, num2):
    if choice == "+":
      return num1 + num2
    elif choice == "-":
      return num1 - num2
    elif choice == "×":
      return num1 * num2
    elif choice == "÷":
      return num1 / num2

demo = gr.Interface(
    fn=operator,
    inputs= [
        gr.Number(label = "Number 1", placeholder="Enter the First Number"),
            gr.Dropdown(
              choices=["+", "-", "×", "÷"],
              label="Select an option",
              value="Option A",  # Default selection
            ),
        gr.Number(label = "Number 2", placeholder="Enter the Second Number"),
            ],
    outputs=gr.Number()
)

demo.launch()
```
