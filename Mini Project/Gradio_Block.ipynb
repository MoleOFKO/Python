```python
#install gradio
!pip install gradio -q

#import gradio
import gradio as gr

#define function
def greet(name):
  return "Hello " + name



#create gradio app with Block
with gr.Blocks() as app:
  gr.Markdown("# Greeding App")
  name = gr.Textbox(label = "Name", placeholder="Enter Your Name"),
  output = gr.Textbox(label = "Greeting")

  button = gr.Button("Greet")

  button.click(
      fn = greet,
      inputs = name,
      outputs = output
  )

#launch
app.launch()
```
