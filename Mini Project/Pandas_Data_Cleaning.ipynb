# **Data Cleaning**
- Data cleaning is the process of finding and fixing missing, duplicate, incorrectm inconsistent, or unnecessary data before performing data analysis or machine learning

# **Reading CSV Data**

```python
from google.colab import drive
drive.mount('/content/drive')

import pandas as pd
#read csv file
df = pd.read_csv("/content/drive/MyDrive/students.csv")
print(df)
```

# **Check Missing Values**
- `**isnull()**` is used to **find missing values**

```python
# print(df.isnull())
print(df.isnull().sum())
```

# **Fill Missing Values with mean, median, mode**

**`fillna()`** - fill missing values
- mean (average value)
- median (middle value) - when data is arranged in ascending or descending order
- mode - is the value that appears most frequently in a dataset

```python
#fill missing values with mean - calculate average of Age column
df["Age"] = df["Age"].fillna(df["Age"].mean())

#fill missing values with median
df["English"] = df["English"].fillna(df["English"].median())

#fill missing values with median
# df["Science"] = df["Science"].fillna(df["Science"].median())

print(df)
```

```python
#remove missing rows
# df = df.dropna()
# print(df)
#remove specific column
df = df.dropna(subset=["Science"])
print(df)
```

```python
#remove
df = df.drop_duplicates()
```

```python
#convert data types
df["Age"] = df["Age"].astype(int)
print(df.dtypes)
```

```python
#find duplicate data
df = df.duplicated().sum()
print(df)
```

```python
#clean column names : sometimes contain unnevessary spaces
df.columns = df.columns.str.strip()
```
